### Q1. Do customers who earn more also spend more?

```python
# data types: income = ?, spending = ?
# tool: ?
# hint: two continuous variables moving together

x = df['income']
y = df['spending']

x = model.add_constant(x)

model = model.OLS(y, x).fit()

predictions = model.predict(x)
print(model.summary())
```

*Your answer:* yes they do seem to spend more if they make more
